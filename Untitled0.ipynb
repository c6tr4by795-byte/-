```python
%%writefile prepare_data.py
import sys
from pathlib import Path
import pandas as pd

OUT_DIR = Path("clean")
TIMEFRAMES = {"M15": "15min", "H1": "1h", "H4": "4h"}

def load(path):
    df = pd.read_csv(path)
    df.columns = [c.strip().lower() for c in df.columns]
    df["time"] = pd.to_datetime(df["timestamp"], unit="ms", utc=True)
    df = df.set_index("time").sort_index()
    return df[["open", "high", "low", "close", "volume"]]

def validate(df):
    r = {"rows_raw": len(df)}
    dup = df.index.duplicated(keep="first")
    r["duplicates_removed"] = int(dup.sum())
    df = df[~dup]
    nan = df.isna().any(axis=1)
    r["nan_rows_removed"] = int(nan.sum())
    df = df[~nan]
    nonpos = (df[["open","high","low","close"]] <= 0).any(axis=1)
    r["non_positive_removed"] = int(nonpos.sum())
    df = df[~nonpos]
    bad = (df["high"] < df[["open","close","low"]].max(axis=1)) | (df["low"] > df[["open","close","high"]].min(axis=1))
    r["inconsistent_ohlc_removed"] = int(bad.sum())
    df = df[~bad]
    step = df.index.to_series().diff()
    gaps = step[(step > pd.Timedelta(minutes=30)) & (step < pd.Timedelta(days=2))]
    r["gaps_over_30min_midweek"] = len(gaps)
    rng = df["high"] - df["low"]
    r["huge_range_candles"] = int((rng > rng.median() * 10).sum())
    r["rows_clean"] = len(df)
    r["first"] = str(df.index[0])
    r["last"] = str(df.index[-1])
    print("=== DATA REPORT ===")
    for k, v in r.items():
        print(f"{k:30s} {v}")
    if len(gaps):
        print("Largest gaps:")
        print(gaps.sort_values(ascending=False).head(10).to_string())
    return df

def resample(df, rule):
    out = df.resample(rule, label="left", closed="left").agg(
        {"open":"first","high":"max","low":"min","close":"last","volume":"sum"})
    return out.dropna(subset=["open"])

def save(df, name):
    OUT_DIR.mkdir(exist_ok=True)
    df.to_parquet(OUT_DIR / f"xauusd_{name}.parquet")

m5 = validate(load(sys.argv[1]))
save(m5, "M5")
for tf, rule in TIMEFRAMES.items():
    d = resample(m5, rule)
    save(d, tf)
    print(f"{tf}: {len(d)} candles")
```

```python
from google.colab import drive; drive.mount("/content/drive")
import pandas as pd, numpy as np
from bisect import bisect_left

DATA = "/content/drive/MyDrive/GOLD-INTELLIGENCE/data/"
SPREAD = 0.30
CUTOFF = "2024-01-01"
TFS = {"M15": (15, [4, 16]), "H1": (60, [4, 12]), "H4": (240, [3, 6])}

def feats(df):
    o, h, l, c = (df[k].values.astype(float) for k in ["open", "high", "low", "close"])
    n = len(df); pc = np.r_[c[0], c[:-1]]
    tr = np.maximum(h - l, np.maximum(abs(h - pc), abs(l - pc)))
    atr = pd.Series(tr).rolling(14).mean().values
    ema = pd.Series(c).ewm(span=50, adjust=False).mean().values
    d = pd.Series(c).diff()
    up = d.clip(lower=0).ewm(alpha=1/14, adjust=False).mean()
    dn = (-d.clip(upper=0)).ewm(alpha=1/14, adjust=False).mean()
    rsi = (100 - 100 / (1 + up / dn.replace(0, np.nan))).fillna(50).values
    trend = np.sign(c - ema); trend[:60] = 0
    body = abs(c - o); rng = np.maximum(h - l, 1e-9)
    uw = h - np.maximum(o, c); lw = np.minimum(o, c) - l
    S = lambda a, k: np.r_[np.full(k, np.nan), a[:-k]]
    ok = ~np.isnan(atr)
    P = {}
    P["engulf"] = ((c > o) & (S(c, 1) < S(o, 1)) & (c >= S(o, 1)) & (o <= S(c, 1)),
                   (c < o) & (S(c, 1) > S(o, 1)) & (c <= S(o, 1)) & (o >= S(c, 1)))
    P["hammer"] = ((lw >= .6*rng) & (body <= .3*rng) & (uw <= .15*rng),
                   (uw >= .6*rng) & (body <= .3*rng) & (lw <= .15*rng))
    dj = (body <= .1*rng) & (rng >= .5*atr)
    P["doji"] = (dj & (S(c,1) < S(c,2)) & (S(c,2) < S(c,3)),
                 dj & (S(c,1) > S(c,2)) & (S(c,2) > S(c,3)))
    big = (rng >= 1.5*atr) & (body >= .7*rng)
    P["bigcandle"] = (big & (c > o), big & (c < o))
    P["3streak"] = ((c > S(c,1)) & (S(c,1) > S(c,2)) & (S(c,2) > S(c,3)),
                    (c < S(c,1)) & (S(c,1) < S(c,2)) & (S(c,2) < S(c,3)))
    for N in (20, 50):
        hh = pd.Series(h).rolling(N).max().shift(1).values
        ll = pd.Series(l).rolling(N).min().shift(1).values
        P[f"break{N}"] = (c > hh, c < ll)
    P["rsi"] = (rsi < 30, rsi > 70)
    P["emacross"] = ((c > ema) & (S(c,1) <= S(ema,1)), (c < ema) & (S(c,1) >= S(ema,1)))
    P = {k: (np.nan_to_num(a).astype(bool) & ok, np.nan_to_num(b).astype(bool) & ok) for k, (a, b) in P.items()}
    return o, c, trend, P

def stats(x):
    n = len(x)
    if n < 2: return n, 0.0, 0.0, 0.0
    m = x.mean(); s = x.std(ddof=1)
    return n, m, (m / (s / np.sqrt(n)) if s > 0 else 0.0), (x > 0).mean() * 100

def thin(pos, H):
    p = pos.tolist(); out = []; k = 0
    while k < len(p):
        out.append(p[k]); k = bisect_left(p, p[k] + H)
    return np.array(out, dtype=int)

def scan():
    res = []
    for tf, (bm, Hs) in TFS.items():
        df = pd.read_parquet(f"{DATA}xauusd_{tf}_final.parquet")
        df = df[df.index < pd.Timestamp(CUTOFF, tz="UTC")]
        assert df.index.max() < pd.Timestamp(CUTOFF, tz="UTC")
        o, c, trend, P = feats(df)
        n = len(df); t = df.index.values
        year = df.index.year.values
        hr = df.index.tz_convert("Europe/London").hour.values
        F = {"all": np.ones(n, bool)}
        if tf != "H4":
            F["asia"] = hr < 7
            F["london"] = (hr >= 8) & (hr < 12)
            F["ny"] = (hr >= 13) & (hr < 17)
        for H in Hs:
            unit = np.full(n, np.nan); unit[:n-H] = c[H:] - o[1:n-H+1]
            gap = np.full(n, np.inf)
            gap[:n-H] = (t[H:] - t[1:n-H+1]).astype("timedelta64[m]").astype(float)
            valid = (gap <= (H - 1) * bm + 120) & ~np.isnan(unit)
            for pat, (bull, bear) in P.items():
                for flip in (False, True):
                    d = bull.astype(int) - bear.astype(int)
                    d[bull & bear] = 0
                    if flip: d = -d
                    filters = list(F.items()) + [("with-trend", (d == trend) & (trend != 0)),
                                                 ("counter-trend", (d == -trend) & (trend != 0))]
                    for fname, fm in filters:
                        pos = np.flatnonzero((d != 0) & fm & valid)
                        if len(pos) < 30: continue
                        pos = thin(pos, H)
                        net = d[pos] * unit[pos] - SPREAD
                        y = year[pos]
                        a = stats(net[y <= 2021]); b = stats(net[y >= 2022])
                        name = f"{tf} {pat} {'fade' if flip else 'follow'} {fname} H{H}"
                        res.append((name, tf, H, *a, *b))
    cols = ["strategy", "tf", "H", "n1", "net1", "t1", "w1", "n2", "net2", "t2", "w2"]
    return pd.DataFrame(res, columns=cols)

def report(R):
    k = len(R)
    p1 = R[(R.n1 >= 100) & (R.net1 > 0) & (R.t1 >= 2.5)]
    p2 = p1[(p1.n2 >= 40) & (p1.net2 > 0) & (p1.t2 >= 1.5)]
    print(f"strategies tested: {k}")
    print(f"passed discovery (2018-21, t>=2.5): {len(p1)}")
    print(f"passed validation too (2022-23, t>=1.5): {len(p2)}")
    print(f"expected false passes by luck: ~{k*0.0062*0.067:.2f}")
    show = ["strategy", "n1", "net1", "t1", "n2", "net2", "t2"]
    print("\nSURVIVORS:")
    print(p2.sort_values("t2", ascending=False).head(15)[show].round(2).to_string(index=False) if len(p2) else "none")
    print("\nTOP 8 by discovery t (to see overfitting):")
    top = R[R.n1 >= 100].sort_values("t1", ascending=False).head(8)
    print(top[show].round(2).to_string(index=False))
    print("\nBEST 5 by validation t (n2>=40):")
    print(R[R.n2 >= 40].sort_values("t2", ascending=False).head(5)[show].round(2).to_string(index=False))

R = scan()
R.to_csv("scan_results.csv", index=False)
report(R)
```

```python
from google.colab import drive; drive.mount("/content/drive")
import pandas as pd, numpy as np

DATA = "/content/drive/MyDrive/GOLD-INTELLIGENCE/data/"
SPREAD = 0.30
CUTOFF = "2024-01-01"
RRS = (1, 2, 3)
LOOK, MSS_WAIT, FILL_WAIT, MAX_HOLD = 6, 24, 24, 96

m5 = pd.read_parquet(DATA + "xauusd_M5_final.parquet")
m5 = m5[m5.index < pd.Timestamp(CUTOFF, tz="UTC")].copy()
assert m5.index.max() < pd.Timestamp(CUTOFF, tz="UTC")
idx = m5.index; n = len(m5)
o, h, l, c = (m5[k].values.astype(float) for k in ["open", "high", "low", "close"])
brk = np.r_[False, np.diff(idx.values) > np.timedelta64(35, "m")]
year = idx.year.values

pc = np.r_[c[0], c[:-1]]
tr = np.maximum(h - l, np.maximum(abs(h - pc), abs(l - pc)))
atr = pd.Series(tr).rolling(14).mean().values

lon = idx.tz_convert("Europe/London")
hr = lon.hour.values + lon.minute.values / 60
m5["ld"] = lon.date
m5["nd"] = (idx.tz_convert("America/New_York") + pd.Timedelta(hours=7)).date
dd = m5.groupby("nd").agg(dh=("high", "max"), dl=("low", "min"))
dd["pdh"] = dd.dh.shift(1); dd["pdl"] = dd.dl.shift(1)
pdh = m5["nd"].map(dd.pdh).values.astype(float); pdl = m5["nd"].map(dd.pdl).values.astype(float)
a = m5[hr < 8].groupby("ld").agg(ah=("high", "max"), al=("low", "min"), cnt=("high", "size"))
a = a[a.cnt >= 60]
ash = m5["ld"].map(a.ah).values.astype(float); asl = m5["ld"].map(a.al).values.astype(float)

h4 = pd.read_parquet(DATA + "xauusd_H4_final.parquet")
h4 = h4[h4.index < pd.Timestamp(CUTOFF, tz="UTC")]
h4["trend"] = np.where(h4.close > h4.close.ewm(span=50, adjust=False).mean(), 1, -1)
L_ = pd.DataFrame({"t": (idx + pd.Timedelta(minutes=5)).as_unit("ns")})
R_ = pd.DataFrame({"known": (h4.index + pd.Timedelta(hours=4)).as_unit("ns"), "trend": h4.trend.values})
trend = pd.merge_asof(L_, R_, left_on="t", right_on="known")["trend"].fillna(0).values

def find_setups(O, H, L, C, lvl, window, tr_ok):
    pos = np.flatnonzero(window)
    days = m5["ld"].values[pos]
    cuts = np.flatnonzero(days[1:] != days[:-1]) + 1
    out = []
    for p in np.split(pos, cuts):
        if len(p) == 0: continue
        lv = lvl[p[0]]
        if np.isnan(lv): continue
        cand = p[L[p] < lv]
        if len(cand) == 0: continue
        S = cand[0]
        if S < LOOK: continue
        Rh = H[S-LOOK+1:S+1].max(); Lw = L[S]; M = -1
        for k in range(S+1, min(S+1+MSS_WAIT, n)):
            if brk[k]: break
            Lw = min(Lw, L[k])
            if C[k] > Rh: M = k; break
        if M < 0: continue
        j = -1
        for jj in range(M-1, S-1, -1):
            if L[jj+1] > H[jj-1]: j = jj; break
        if j < 0: continue
        E = (H[j-1] + L[j+1]) / 2
        am = atr[M]
        if np.isnan(am): continue
        SL = Lw - 0.1 * am; risk = E - SL
        if risk < 0.5 * am: continue
        f = -1
        for k in range(M+1, min(M+1+FILL_WAIT, n)):
            if brk[k]: break
            if L[k] <= E: f = k; break
        if f < 0: continue
        out.append((year[S], f, E, SL, risk, tr_ok[S]))
    return out

def sim(f, E, SL, risk, RR, H, L, C):
    cost = SPREAD / risk
    if L[f] <= SL: return -1 - cost
    tp = E + RR * risk
    end = min(f + MAX_HOLD, n - 1)
    for k in range(f+1, end+1):
        if brk[k]: return (C[k-1] - E) / risk - cost
        if L[k] <= SL: return -1 - cost
        if H[k] >= tp: return RR - cost
    return (C[end] - E) / risk - cost

def stats(x):
    x = np.asarray(x, float); k = len(x)
    if k < 2: return k, 0.0, 0.0, 0.0
    mu = x.mean(); s = x.std(ddof=1)
    return k, mu, (mu / (s / np.sqrt(k)) if s > 0 else 0.0), (x > 0).mean() * 100

rows = []
for kind in ("PDH/PDL", "ASIA"):
    win = ((hr >= 7) & (hr < 17)) if kind == "PDH/PDL" else ((hr >= 8) & (hr < 12))
    lv_long = pdl if kind == "PDH/PDL" else asl
    lv_short = pdh if kind == "PDH/PDL" else ash
    S_long = find_setups(o, h, l, c, lv_long, win, trend == 1)
    S_short = find_setups(-o, -l, -h, -c, -lv_short, win, trend == -1)
    for use_tr in (False, True):
        for RR in RRS:
            res = []
            for side, (O, H, L, C), S in ((1, (o, h, l, c), S_long), (-1, (-o, -l, -h, -c), S_short)):
                for (y, f, E, SL, risk, ok) in S:
                    if use_tr and not ok: continue
                    res.append((y, sim(f, E, SL, risk, RR, H, L, C)))
            r = np.array(res) if res else np.zeros((0, 2))
            a1 = stats(r[r[:, 0] <= 2021, 1]); a2 = stats(r[r[:, 0] >= 2022, 1])
            rows.append((f"{kind} {'+H4trend' if use_tr else 'no-filter'} RR{RR}", *a1, *a2))

R = pd.DataFrame(rows, columns=["setup", "n1", "R1", "t1", "w1", "n2", "R2", "t2", "w2"])
print("SWEEP -> MSS -> FVG limit entry (R = profit in units of risk, after spread)")
print(R.round(2).to_string(index=False))
p = R[(R.n1 >= 100) & (R.R1 > 0) & (R.t1 >= 2.0) & (R.n2 >= 40) & (R.R2 > 0) & (R.t2 >= 1.5)]
print(f"\ntests: {len(R)} | passed both gates: {len(p)}")
print(p[["setup", "n1", "R1", "t1", "n2", "R2", "t2"]].round(2).to_string(index=False) if len(p) else "none passed")
```

```python
from google.colab import drive; drive.mount("/content/drive")
import pandas as pd, numpy as np
from bisect import bisect_left

DATA = "/content/drive/MyDrive/GOLD-INTELLIGENCE/data/"
SPREAD = 0.30
CUTOFF = "2024-01-01"
TFS = {"M15": (15, [4, 16]), "H1": (60, [4, 12]), "H4": (240, [3, 6])}

def feats(df, tf):
    o, h, l, c = (df[k].values.astype(float) for k in ["open", "high", "low", "close"])
    n = len(df); pc = np.r_[c[0], c[:-1]]
    tr = np.maximum(h - l, np.maximum(abs(h - pc), abs(l - pc)))
    atr = pd.Series(tr).rolling(14).mean().values
    ema = pd.Series(c).ewm(span=50, adjust=False).mean().values
    d = pd.Series(c).diff()
    up = d.clip(lower=0).ewm(alpha=1/14, adjust=False).mean()
    dn = (-d.clip(upper=0)).ewm(alpha=1/14, adjust=False).mean()
    rsi = (100 - 100 / (1 + up / dn.replace(0, np.nan))).fillna(50).values
    trend = np.sign(c - ema); trend[:60] = 0
    body = abs(c - o); rng = np.maximum(h - l, 1e-9)
    uw = h - np.maximum(o, c); lw = np.minimum(o, c) - l
    S = lambda a, k: np.r_[np.full(k, np.nan), a[:-k]]
    ok = ~np.isnan(atr)
    P = {}
    P["engulf"] = ((c > o) & (S(c, 1) < S(o, 1)) & (c >= S(o, 1)) & (o <= S(c, 1)),
                   (c < o) & (S(c, 1) > S(o, 1)) & (c <= S(o, 1)) & (o >= S(c, 1)))
    P["hammer"] = ((lw >= .6*rng) & (body <= .3*rng) & (uw <= .15*rng),
                   (uw >= .6*rng) & (body <= .3*rng) & (lw <= .15*rng))
    dj = (body <= .1*rng) & (rng >= .5*atr)
    P["doji"] = (dj & (S(c,1) < S(c,2)) & (S(c,2) < S(c,3)),
                 dj & (S(c,1) > S(c,2)) & (S(c,2) > S(c,3)))
    big = (rng >= 1.5*atr) & (body >= .7*rng)
    P["bigcandle"] = (big & (c > o), big & (c < o))
    P["3streak"] = ((c > S(c,1)) & (S(c,1) > S(c,2)) & (S(c,2) > S(c,3)),
                    (c < S(c,1)) & (S(c,1) < S(c,2)) & (S(c,2) < S(c,3)))
    for N in (20, 50):
        hh = pd.Series(h).rolling(N).max().shift(1).values
        ll = pd.Series(l).rolling(N).min().shift(1).values
        P[f"break{N}"] = (c > hh, c < ll)
    P["rsi"] = (rsi < 30, rsi > 70)
    P["emacross"] = ((c > ema) & (S(c,1) <= S(ema,1)), (c < ema) & (S(c,1) >= S(ema,1)))
    for k in (5, 10):
        hmax = pd.Series(h).rolling(2*k+1, center=True).max().values
        lmin = pd.Series(l).rolling(2*k+1, center=True).min().values
        fh = np.full(n, np.nan); fl = np.full(n, np.nan)
        j = np.flatnonzero(h == hmax); j = j[j + k < n]; fh[j + k] = h[j]
        j = np.flatnonzero(l == lmin); j = j[j + k < n]; fl[j + k] = l[j]
        res = pd.Series(fh).ffill().values; sup = pd.Series(fl).ffill().values
        prox = .25 * atr
        P[f"sr{k}_bounce"] = ((l <= sup + prox) & (c > sup) & (c > o), (h >= res - prox) & (c < res) & (c < o))
        P[f"sr{k}_break"] = ((c > res) & (S(c,1) <= res), (c < sup) & (S(c,1) >= sup))
        P[f"sr{k}_fake"] = ((l < sup) & (c > sup), (h > res) & (c < res))
    for s in (10, 50):
        base = np.floor(o / s) * s; up = base + s; dn = base
        P[f"rn{s}_reject"] = ((l <= dn) & (c > dn), (h >= up) & (c < up))
        P[f"rn{s}_break"] = ((c > up), (c < dn))
    if tf == "H1":
        hrs = df.index.tz_convert("Europe/London").hour.values
        for hh in range(24):
            P[f"enter@{hh:02d}"] = (hrs == (hh - 1) % 24, np.zeros(n, bool))
    P = {k: (np.nan_to_num(a).astype(bool) & ok, np.nan_to_num(b).astype(bool) & ok) for k, (a, b) in P.items()}
    return o, c, trend, P

def stats(x):
    n = len(x)
    if n < 2: return n, 0.0, 0.0, 0.0
    m = x.mean(); s = x.std(ddof=1)
    return n, m, (m / (s / np.sqrt(n)) if s > 0 else 0.0), (x > 0).mean() * 100

def thin(pos, H):
    p = pos.tolist(); out = []; k = 0
    while k < len(p):
        out.append(p[k]); k = bisect_left(p, p[k] + H)
    return np.array(out, dtype=int)

def scan():
    res = []
    for tf, (bm, Hs) in TFS.items():
        df = pd.read_parquet(f"{DATA}xauusd_{tf}_final.parquet")
        df = df[df.index < pd.Timestamp(CUTOFF, tz="UTC")]
        assert df.index.max() < pd.Timestamp(CUTOFF, tz="UTC")
        o, c, trend, P = feats(df, tf)
        n = len(df); t = df.index.values
        year = df.index.year.values
        hr = df.index.tz_convert("Europe/London").hour.values
        F = {"all": np.ones(n, bool)}
        if tf != "H4":
            F["asia"] = hr < 7
            F["london"] = (hr >= 8) & (hr < 12)
            F["ny"] = (hr >= 13) & (hr < 17)
        for H in Hs:
            unit = np.full(n, np.nan); unit[:n-H] = c[H:] - o[1:n-H+1]
            gap = np.full(n, np.inf)
            gap[:n-H] = (t[H:] - t[1:n-H+1]).astype("timedelta64[m]").astype(float)
            valid = (gap <= (H - 1) * bm + 120) & ~np.isnan(unit)
            for pat, (bull, bear) in P.items():
                for flip in (False, True):
                    d = bull.astype(int) - bear.astype(int)
                    d[bull & bear] = 0
                    if flip: d = -d
                    filters = list(F.items()) + [("with-trend", (d == trend) & (trend != 0)),
                                                 ("counter-trend", (d == -trend) & (trend != 0))]
                    if pat.startswith("enter@"): filters = filters[:1]
                    for fname, fm in filters:
                        pos = np.flatnonzero((d != 0) & fm & valid)
                        if len(pos) < 30: continue
                        pos = thin(pos, H)
                        net = d[pos] * unit[pos] - SPREAD
                        y = year[pos]
                        a = stats(net[y <= 2021]); b = stats(net[y >= 2022])
                        name = f"{tf} {pat} {'fade' if flip else 'follow'} {fname} H{H}"
                        res.append((name, tf, H, *a, *b))
    cols = ["strategy", "tf", "H", "n1", "net1", "t1", "w1", "n2", "net2", "t2", "w2"]
    return pd.DataFrame(res, columns=cols)

def report(R):
    R = R.copy()
    k = len(R)
    p1 = R[(R.n1 >= 100) & (R.net1 > 0) & (R.t1 >= 2.5)]
    p2 = p1[(p1.n2 >= 40) & (p1.net2 > 0) & (p1.t2 >= 1.5)]
    print(f"strategies tested: {k}")
    print(f"passed discovery (2018-21, t>=2.5): {len(p1)}")
    print(f"passed validation too (2022-23, t>=1.5): {len(p2)}")
    print(f"expected false passes by luck: ~{k*0.0062*0.067:.2f}")
    fam = lambda s: ("time" if "enter@" in s else "classical S/R" if " sr" in s else "round numbers" if " rn" in s else "candles/breakouts/RSI")
    R["family"] = R.strategy.map(fam)
    print("\nBY FAMILY (tested / passed discovery / passed both):")
    for f_, g in R.groupby("family"):
        q1 = g[(g.n1 >= 100) & (g.net1 > 0) & (g.t1 >= 2.5)]
        q2 = q1[(q1.n2 >= 40) & (q1.net2 > 0) & (q1.t2 >= 1.5)]
        print(f"  {f_}: {len(g)} / {len(q1)} / {len(q2)}")
    show = ["strategy", "n1", "net1", "t1", "n2", "net2", "t2"]
    print("\nSURVIVORS:")
    print(p2.sort_values("t2", ascending=False).head(15)[show].round(2).to_string(index=False) if len(p2) else "none")
    print("\nTOP 8 by discovery t (to see overfitting):")
    top = R[R.n1 >= 100].sort_values("t1", ascending=False).head(8)
    print(top[show].round(2).to_string(index=False))
    print("\nBEST 5 by validation t (n2>=40):")
    print(R[R.n2 >= 40].sort_values("t2", ascending=False).head(5)[show].round(2).to_string(index=False))

R = scan()
R.to_csv("scan_results.csv", index=False)
report(R)
```
